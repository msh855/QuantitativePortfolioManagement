```python
## IMPORT lIBRARIES

import warnings
warnings.filterwarnings('ignore')


import os 
import pandas as pd 

# for clustering libraries 
from tslearn.clustering import TimeSeriesKMeans, silhouette_score
from tslearn.preprocessing import TimeSeriesScalerMinMax
from tslearn.preprocessing import TimeSeriesScalerMeanVariance

from datetime import date, datetime

# to smooth series 
from tsmoothie.smoother import * 

# for plottign 
import matplotlib.pyplot as plt

# to download data
from yahoofinancials import YahooFinancials
```

```python
# set working directory 
os.chdir("/Users/safishajjouz/GitHub/myPortfolioManagement/files")
```

```python
# Download data 
# Set dates 
start_date = '1993-01-01'
date_end = date.today()
date_end = date_end.strftime("%Y-%m-%d")


# # Download Data
tickers = ['SPY','^VIX'] # VIX

yahoo_financials = YahooFinancials(tickers)
data = yahoo_financials.get_historical_price_data(start_date=start_date, 
                                                  end_date=date_end, 
                                                  time_interval='daily')

data_list = []
for tik in tickers:
    prices = pd.DataFrame.from_dict(data[tik]['prices'])
    prices['yahoo_ticker'] = tik
    # fix faulty yahoo data that jumps 100x
    jumps_up   = prices['adjclose'] / prices['adjclose'].shift() >  50
    jumps_down = prices['adjclose'] / prices['adjclose'].shift() < .02
    correction_factor = 100.**(jumps_down.cumsum() - jumps_up.cumsum())
    prices['adjclose'] *= correction_factor
    print(f"Fixed {sum(correction_factor != 1)}/{len(data)} for ticker {tik}"
          f" (min: {prices['adjclose'].min()}, max: {prices['adjclose'].max()})")
    data_list.append(prices)
    
df = pd.concat(data_list)
df = df.rename(columns={'formatted_date':'Date'})
df = df.drop('date', axis = 1)
df = df.set_index('Date')
```

```python
df = df[['adjclose', 'yahoo_ticker']]
```

```python
df_cleaned = df.pivot_table(index=["Date"], 
                    columns='yahoo_ticker', 
                    values='adjclose')
df_cleaned = df_cleaned.rename(columns={'^VIX':"VIX"})
df_cleaned
```

```python
df_cleaned[['SPY']].plot()
df_cleaned[['VIX']].plot()
```

```python
# Fill Missining Values 
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
df_cleaned.index = pd.to_datetime(df_cleaned.index, format='%Y/%m/%d')
df_cleaned.interpolate(method='time', inplace=True)

# smooth Series 
data_series = df_cleaned.transpose().to_numpy()
df_cleaned_smoothed = LowessSmoother(smooth_fraction=0.1, iterations=1)  # higher values for smooth fraction more smooth (trend)
df_cleaned_smoothed.smooth(data_series)

# generate smoothed DataFrame 
df_cleaned_sm = df_cleaned.copy()
for i in range(len(df_cleaned.columns)):
    df_cleaned_sm.iloc[:,i] = df_cleaned_smoothed.smooth_data[i]
```

```python
plt.figure(figsize=(5, 5))
plt.plot(df_cleaned['VIX'])
plt.plot(df_cleaned_sm['VIX'])
plt.show()

plt.figure(figsize=(5, 5))
plt.plot(df_cleaned['SPY'])
plt.plot(df_cleaned_sm['SPY'])
plt.show()
```

```python
from tsmoothie.smoother import *
temp = temp[temp.index=='VIX']

window_shape = 60
smoother = WindowWrapper(LowessSmoother(smooth_fraction=0.6, iterations=1), 
                         window_shape=window_shape)
smoother.smooth(temp)
low, up = smoother.get_intervals('prediction_interval') # only for visualization
```

```python
## SCALE SMOOTHED WINDOWS ###
from tslearn.clustering import TimeSeriesKMeans
from tslearn.preprocessing import TimeSeriesScalerMinMax

smooth_scaled = TimeSeriesScalerMinMax().fit_transform(smoother.Smoother.smooth_data)
smooth_scaled.shape

## CLUSTER SMOOTHED WINDOWS ###
smooth_km = TimeSeriesKMeans(n_clusters=4, metric="dtw", max_iter=10, random_state=33)
smooth_km.fit(smooth_scaled)
### PLOT CLUSTER CENTERS WITH SAMPLES ###
colors = {0:'red',1:'blue',2:'green',3:'orange'}

for c in range(smooth_km.n_clusters):
    
    #plt.plot(np.squeeze(smooth_scaled[smooth_km.labels_ == c],-1).T, c=colors[c], alpha=0.05)
    plt.plot(np.squeeze(smooth_km.cluster_centers_, -1)[c], c=colors[c], linewidth=5)
    plt.title(f"cluster {c}")
    
    plt.show()
```

```python
### PLOT CLUSTER ON THE RAW TIME SERIES ###
import plotly.express as px

plt.figure(figsize=(16,8))
plt.plot(smoother.Smoother.data[:,-1], c='black', alpha=0.3)
plt.scatter(range(len(smoother.Smoother.data[:,-1])), smoother.Smoother.data[:,-1], 
            c=[colors[c] for c in smooth_km.labels_])

plt.xticks(range(0,len(df.index.unique())-window_shape, 100), 
           df.index.unique()[window_shape::100], rotation=70)
plt.ylabel('VIX')
np.set_printoptions(False)
```
