```python
import quandl
quandl.ApiConfig.api_key = 'fhbmNKX6oNP7PpFuZJNo'

from fredapi import Fred
fred_api_key = 'cc628b51e21828ae6b98c06f4eef6714'
fred = Fred(api_key=fred_api_key)

from datetime import date, datetime

import pandas as pd
```

```python
# Set dates 
start_date = '1993-01-01'
date_end = date.today()
date_end = date_end.strftime("%Y-%m-%d")
```

```python
# FRED St. Louis Data 
fred_series = ["WILLLRGCAPGRPR", 
               "WILLLRGCAPPR",
               "WILLLRGCAPVALPR",
               'VIXCLS']

df_fred = pd.DataFrame()
for series in fred_series:
    s = fred.get_series(series, observation_start=start_date, observation_end=date_end)
    s.name = series
    df_fred = df_fred.join(s, how='outer')
```

```python
df_fred[['VIXCLS']].plot()
```

```python
# Quandl Data 
fred_series = ["DGS10", "T5YIE", "DCOILWTICO", "DEXUSEU"]
fred_stamp = "FRED/"
fred_ids = [fred_stamp + series  for series in fred_series ]


ids = fred_ids + ["LBMA/GOLD"] +  ["USTREASURY/YIELD"]

data = quandl.get(ids, start_date = start_date, end_date= date_end)
data
```
