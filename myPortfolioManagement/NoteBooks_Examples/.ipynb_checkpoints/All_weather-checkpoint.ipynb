```python
# All-weather
start_trading_date = "2020-11-24"
tickers = ['EEM', 'VNQ', 'MDY', 'SLY', 'SPY', 'EFA', 'TIP', 'AGG', 'DJP', 'BIL']
df_prices = get_stock_prices(tickers, wide_format=True)
df_prices = df_prices[df_prices.index >= start_trading_date]
```

```python
def rebase(df_prices):
    return ffn.core.rebase(df_prices)
```

```python
rebase(df_prices).plot(figsize = (10,10))
```

```python
ret = df_prices.pct_change()
weights = equal_weight_portfolio(ret)
all_weather_ret = calculate_portfolio_returns(ret, weights.reset_index(), portfolio_name='All_weather')
all_weather_ret
```

```python
all_weather_ret.index = pd.to_datetime(all_weather_ret.index)
```
